```python
import numpy as np
from sklearn.decomposition import PCA


#(Ax = λx)

print("--- Raw Eigenvectors & Eigenvalues ---")


A = np.array([[4, -2],
              [1,  1]])

# 2. Calculate Eigenvalues (λ) and Eigenvectors (x) using NumPy
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A:\n", A)
print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)



print("\n--- Dimensionality Reduction (PCA) ---")

# 1. Simulate a massive dataset (e.g., 1,000 houses, 500 features/columns each)
# This creates a matrix of size 1000 x 500
massive_dataset = np.random.rand(1000, 500)
print(f"Original dataset shape: {massive_dataset.shape}")

# 2. Initialize PCA to compress the 500 dimensions down to 10
# Under the hood, this calculates the eigenvalues and keeps the top 10
pca = PCA(n_components=10)

# 3. Compress the data
compressed_dataset = pca.fit_transform(massive_dataset)

print(f"Compressed dataset shape: {compressed_dataset.shape}")

# 4. Check how much of the original "truth" (variance) was kept
variance_kept = sum(pca.explained_variance_ratio_) * 100
print(f"Data compressed to 10 columns. Information retained: {variance_kept:.2f}%")
```

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import kaggle


kaggle.api.dataset_download_files('camnugent/california-housing-prices', path='.', unzip=True)
df = pd.read_csv('housing.csv')

numeric_df = df.select_dtypes(include=['float64', 'int64']).dropna()

print("BEFORE PCA:")
print(numeric_df.columns.tolist())
print("-" * 50)

scaler = StandardScaler()
scaled_data = scaler.fit_-transform(numeric_df)

pca = PCA(n_components=3)
compressed_data = pca.fit_transform(scaled_data)

# 4. Convert the raw math array back into a Pandas DataFrame
# We have to name the columns ourselves because they are brand new math concepts!
pca_df = pd.DataFrame(compressed_data, columns=['Component_1', 'Component_2', 'Component_3'])

# --- THE "AFTER" COLUMNS ---
print("\nAFTER PCA (3 Artificial Columns):")
print(pca_df.columns.tolist())
```
